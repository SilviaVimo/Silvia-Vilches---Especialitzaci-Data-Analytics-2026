Primero de todo, tenemos que pedir los datos que necesitamos de cada candidato
para luego poder obtener la información que RRHH nos ha solicitado.
Esta información, debemos guardarla en un diccionario para poder efectuar los calculos.


---


    
**FUNCIÓN** obtener_datos:
1. Pedir nombre y apellidos (texto - str)
2. Pedir número de teléfono de contacto (texto- str)
3. Pedir los años de experiencia (num entero ≥ 0)
4. Pedir la puntuación en catalán (num entero 0-3)
5. Pedir la puntuación en Excel (num entero 0-3)
6. Pedir el título académico (num entero 0-3)
7. Devolver un diccionario con todos los datos


```python
def obtener_datos(numero_candidato):

  print(f"CANDIDATO {numero_candidato} DE 10")
  nombre = input("Introduce tu nombre y primer apellido: ")
  telefono = input("Introduce tu número de teléfono: ") # definimos que sea un numero entero

  # que nos de un error cuando el valor sea negativo
  try:
    experiencia = int(input("Introduce tus años de experiencia: "))
  except ValueError:
    experiencia < 0

  # que nos de un error cuando el valor no esté entre 0 y 3
  try:
    catalan = int(input("Introduce tu puntuación en catalán (0 = sin conocimientos // 1 = hasta el nivel B1 // 2 = nivel B2 // 3 = C1 o superior): "))
  except ValueError:
    catalan < 0 or catalan > 3

  try:
    excel = int(input("Introduce tu puntuación en Excel (0 = nivel básico // 1 = uso de funciones // 2 = tablas dinámicas // 3 = Visual Basic y macros):  "))
  except ValueError:
    excel < 0 or excel > 3

  try:
    titulo = int(input("Introduce tu título académico (0 = ESO // 1 = Título de formación profesional // 2 = Universidad // 3 = Máster o título superior) "))
  except ValueError:
    titulo < 0 or titulo > 3

  # que guarde los datos
  datos = {
        "nombre": nombre,
        "telefono": telefono,
        "experiencia": experiencia,
        "catalan": catalan,
        "excel": excel,
        "titulo": titulo
         }
  return datos
```

Comprobamos que todos los datos introducidos manualmente por los candidatos sean válidos, estén completos y que estén dentro del rango que RRHH nos ha dado.
Si todo el correcto, necesitamos que estos datos se guarden. Si en algún momento se introducen datos que no complen los requisitos dado, saldrá en la pantalla un error.


---




**FUNCIÓN** validar_datos(datos del candidato obtenidos en el código anterior):
  1. Comprobar que "nombre" no esté vacío
  2. Comprobar que "apellidos" no esté vacío
  3. Comprobar que "telefono no esté vacío
  4. Comprobar que "experiencia" sea >= 0
  5. Comprobar que catalan, excel y titulo estén entre 0 y 3
  6. Si todo es válido →  True
  6. Si hay un error → muestra un mensaje y devuelve False

```python
def validar_datos(datos):

  if not datos["nombre"]:
        print("ERROR: Este campo no puede estar vacio")
        return False

  if not datos["telefono"]:
        print("ERROR: Este campo no puede estar vacio")
        return False

  if datos["experiencia"] < 0:
        print("ERROR: Los años de experiencia no pueden ser negativos")
        return False

  if datos["catalan"] < 0 or datos["catalan"] > 3:
        print("ERROR: La puntuación de catalán debe estar entre 0 y 3")
        return False

  if datos["excel"] < 0 or datos["excel"] > 3:
        print("ERROR: La puntuación de excel debe estar entre 0 y 3")
        return False

  if datos["titulo"] < 0 or datos["titulo"] > 3:
        print("ERROR: La puntuación del titulo debe estar entre 0 y 3")
        return False

  return True
```

Calcular la puntuación total de cada candidato con la fórmula que RRHH nos ha pedido:
puntuación total = años de experiencia * 1.5 + nivel catalán + nivel excel + título


---



**FUNCIÓN** calcular_puntuacion(con los datos obtenidos del candidato):
  1. Calcular años_experiencia = experiencia * 1,5
  2. Sumar catalan + excel + titulo
  3. puntuacion_total = años_experiencia + catalan + excel + titulo
  4. Devolver puntuacion_total y que quede guardada.

```python
def calcular_puntuacion(datos):
  años_experiencia = datos["experiencia"] * 1.5
  puntuacion_total = años_experiencia + datos["catalan"] + datos["excel"] + datos["titulo"]
  return puntuacion_total
```

Ahora necesitamos clasificar a los candidatos inscritos según los umbrales establecidos por la empresa.


---


**FUNCIÓN** clasificar_candidato(puntuacion_total obtenida de la función anterior):
  1. Si la puntuación total >= 12  → devolver "Prioridad alta"
  2. Si la puntuación total 8 <= o < 12 → devolver "Candidato adecuado"
  3. Si la puntuación total < 8   → devolver "No prioritario"

  Utilizamos condicionales para generar esta clasificación.

```python
def clasificar_candidato(puntuacion_total):
    if puntuacion_total >= 12:
      return "Prioridad alta", 1          # le damos un orden, orden 1 para después poder ordenar
    elif puntuacion_total <= 8 and puntuacion_total < 12:
      return "Candidato adecuado", 2      # orden 2
    else:
      return "No prioritario", 3          # orden 3
```

Necesitamos, para más tarde crear un informe, mostrar los resultados obtenidos de cada candidato una vez rellenado el formulario.


---


**FUNCIÓN** mostrar_individual(datos_candidato, puntuacion_total, clasificaciónde la función anterior):
  1. Mostrar nombre y apellidos del candidato
  2. Mostrar puntuación total que ya ha sido calculada
  3. Mostrar clasificación recibida por la función anterior

```python
def mostrar_individual(datos, puntuacion_total, clasificacion):
    print(f"Candidato/a: {datos["nombre"]}")
    print(f"Puntuación:  {puntuacion_total}")
    print(f"Categoria: {clasificacion}")
```

Por último, tenemos que generar un informe final con el nombre, apellidos, teléfono, puntuación y prioridad de todos los candidatos inscritos para que RRHH pueda llamarlos sin necesidad que recurrir a nosotros para obtener los datos.
Estos candidatos deben estar ordenados por puntuación de mayor a menor, por lo tanto, hay qué definir cómo ordenarlos y tener en cuenta los empates.



---


**FUNCIÓN** mostrar_informe(con todos los candidatos que se han inscrito, hasta un máximo de 10 pax):
  1. Obtener los datos guardados
  2. Definir el orden de "*Prioridad Alta*" a "*No prioritario*"
  3. Ordenar los 10 candidatos (Tener en cuenta los empates)
  4. Mostrar los datos que se piden de cada candidato:  Nombre, teléfono, puntuación y tipo de prioridad.
  5. Trasnformar en una tabla (dataframe) para que sea más fácil de visualizar.

```python
def mostrar_informe(lista_candidatos):
  import pandas as pd # para luego poder transfromar en un dataframe el listado de los candidatos

  l = len(lista_candidatos)
  for i in range(l - 1):
        for j in range(l - 1 - i):
            candidato_a = lista_candidatos[j]
            candidato_b = lista_candidatos[j + 1]

            # Comparar por prioridad primero
            if candidato_a["orden"] > candidato_b["orden"]:
                lista_candidatos[j] = candidato_b
                lista_candidatos[j + 1] = candidato_a
            elif candidato_a["orden"] == candidato_b["orden"]:
                # Empate en prioridad: mayor puntuación primero
                if candidato_a["puntuacion"] < candidato_b["puntuacion"]:
                    lista_candidatos[j] = candidato_b
                    lista_candidatos[j + 1] = candidato_a

    # Mostrar el informe
  print("\n")
  print("-------------------------------------")
  print("INFORME FINAL – PROCESO DE SELECCIÓN")
  print("-------------------------------------")

  lista_candidatos_df = pd.DataFrame(lista_candidatos)
  display(lista_candidatos_df)

  print("-------------------------------------")
  print("Fin del informe.")
  print("-------------------------------------")

```

  Por último definimos la **funció principal** que combina todas las funciones anteriores y ejecutará el programa que nos han pedido de una manera eficaz.


---
**FUNCIÓN** main_seleccion():
  1. Crear una lista de candidatos que esté vacia para después añadir los canditatos inscritos.
  2. Añadir un limite a la lista, máximo 10.
  3. Obtener los datos de los candidatos
  4. Validad los datos: Si todos están OK, continuar. Si no son válido, volver a empezar.
  5. Calculamos la puntuación de cada candidato.
  6. Clasificarlos por prioridad.
  7. Mostrar los datos individuales.
  8. Los añadimos a la lsita creada en el paso 1.
  9. Ordenamos la lista.
  10. Cuando la lista haya llegado a su límite, esté con exactamente 10 candidatos, mostrar el informe final con los datos que nos han pedido.


```python
def main_selection():

    print("-----------------------------------------------------")
    print("SISTEMA DE INSCRIPCIÓN – OFERTA DE EMPLEO INTERNA")
    print("Solo se aceptan las primeras 10 solicitudes.")
    print("-----------------------------------------------------")

    lista_candidatos = []
    numero_candidato = 1

    while numero_candidato <= 10:

        # 1. Obtener datos en bruto
        datos = obtener_datos(numero_candidato)

        # 2. Validar datos; si hay error, repetir sin contar la inscripción
        valido = validar_datos(datos)
        if valido == False:
            print("Por favor, vuelve a introducir los datos correctamente.")
            continue  # Repetir sin sumar el valor al contador

        # 3. Calcular puntuación
        puntuacion = calcular_puntuacion(datos)

        # 4. Clasificar candidato
        clasificacion, orden = clasificar_candidato(puntuacion)

        # 5. Mostrar resultado individual
        mostrar_individual(datos, puntuacion, clasificacion)

        # 6. Guardar candidato en la lista
        candidato = {
            "nombre": datos["nombre"],
            "telefono": datos["telefono"],
            "puntuacion": puntuacion,
            "clasificacion": clasificacion,
            "orden": orden
        }
        lista_candidatos.append(candidato)

        numero_candidato = numero_candidato + 1

    # 7. Mostrar informe final con los 10 candidatos
    print("Se han registrado 10 candidatos.")
    mostrar_informe(lista_candidatos)

```

```python
#COMPROBAMOS EL PROGRAMA
main_selection()
```

# DEFINIR LAS FUNCIONES que recibirá **procesar_clientes**

**FUNCIÓN** obtener_segmento_y_factor():

Nos devolverá el segmento, el mensaje y el factor_codigo de cada cliente.

Recibe un código de comportamiento (0-4) y retorna un diccionario con:

1. factor_codigo: peso numérico del comportamiento

2. segmento: nombre del segmento de marketing

3. mensaje: mensaje publicitario personalizado


```python
def obtener_segmento_y_factor(codigo_comportamiento):
    return tabla_comportamiento[codigo_comportamiento]
```

**FUNCIÓN** obtener_factor_edad():

Nos devolverá el factor según el grupo de edad del cliente.

Recibe el grupo de edad del cliente y nos retorna el factor multiplicador correspondiente.

```python
def obtener_factor_edad(grupo_edad):
    return factor_edad[grupo_edad]
```

**FUNCIÓN** calcular_riesgo():

Calcula el riesgo final teniendo el cuenta la *Fórmula de riesgo final*.

Fórmula de riesgo final:
**texto en negrita**
riesgo_final = factor_código * factor_edad.


```python
def calcular_riesgo(factor_codigo, factor_edad):
    return round(factor_codigo * factor_edad, 3)
```

**FUNCIÓN** procesar_clientes():
1. Recibe la lista de nuestros clientes
2. Hacemos un bucle, analizará cada cliente del dicc de clientes.
3. De cada cliente, usaremos las funciones definidas antes: obtener segmento y factor, obtener factor edad y finalmente calcular su riesgo.
4. Una vez hecho esto, añadirá esta información como un diccionario a la lista creada en el paso 1.
5. Pedimos que nos devuelva un diccionario de diccionarios con segmento, mensaje, factores y riesgo_final.
.

```python
def procesar_clientes(lista_clientes):

    clientes_procesados = {}

    for cliente in lista_clientes:
        id = cliente["id_cliente"]

        # Obtener datos de comportamiento (sin condicionales)
        info_comportamiento = obtener_segmento_y_factor(cliente["codigo_comportamiento"])
        factor_codigo = info_comportamiento["factor_codigo"]
        segmento = info_comportamiento["segmento"]
        mensaje = info_comportamiento["mensaje"]

        # Obtener factor de edad (sin condicionales)
        factor_edad = obtener_factor_edad(cliente["grupo_edad"])

        # Calcular riesgo final
        riesgo_final = calcular_riesgo(factor_codigo, factor_edad)

        # Construir registro completo
        clientes_procesados[id] = {
            "id_cliente": id,
            "nombre": cliente["nombre"],
            "edad": cliente["edad"],
            "grupo_edad": cliente["grupo_edad"],
            "codigo_comportamiento":cliente["codigo_comportamiento"],
            "segmento": segmento,
            "mensaje": mensaje,
            "factor_codigo": factor_codigo,
            "factor_edad": factor_edad,
            "riesgo_final": riesgo_final
        }

    return clientes_procesados
```

**FUNCIÓN** split_datos():
Nos devolverá la lista de pacientes en un diccionario con cada uno de ellos separados.
1. Dividir la lista por ","
2. Crear un diccionario con cada paciente
3. Convertir el diccionario en un dataframe para que sea mas facil visualizar y separar datos.

```python
def split_datos(pacientes):
    lista_pacientes = []
    for p in pacientes:
        partes = p.split("|")
        lista_pacientes.append({
            "nombre":   partes[0].strip(),
            "edad":     int(partes[1].strip()),
            "sintomas": partes[2].strip()
        })
    return lista_pacientes
```

**FUNCIÓN** limpiar_datos:

Recibe el diccionario de pacientes.
1. Separar los sintomas por las comas.

**FUNCIÓN** desglosar_pacientes():

```python
def desglosar_pacientes(lista_pacientes):
  pacientes_desglosados = {}

  # Recorrer la lista y asignar cada diccionario usando el nombre como clave
  for paciente in lista_pacientes:
      nombre = paciente["nombre"]
      pacientes_desglosados[nombre] = paciente

  return pacientes_desglosados
```

```python
def calcular_riesgo(paciente, niveles_riesgo):
    suma_riesgo = 0
  # Recorrer cada paciente en la lista

    for sintoma in paciente["sintomas"]:
        suma_riesgo += niveles_riesgo.get(sintoma, 0)
    return suma_riesgo
```

```python
def obtener_random(var1, var2, var3, var4, longitud):
  caracteres = var1 + var2 + var3 + var4

  contraseña = []

  for i in range(longitud):
    caracteres_random = secrets.choice(caracteres)
    contraseña.append(caracteres_random)
    contraseña_final = "".join(contraseña)
  return contraseña_final
```

```python
def historico_dataframe(archivo):

  #convertir a dataframe
  archivo = pd.read_csv("historic_partits.txt", sep="\t")

  # definir las columnas
  archivo.columns = ["equipo_local", "resultado", "equipo_visitante"]

  # dividir la columna "resultados" en 2
  archivo[["goles_local", "goles_visitante"]] = archivo["resultado"].str.split("-", expand=True)

  #eliminamos la columna de "resultados"
  df = archivo.drop(columns=["resultado"])

  # ordenamos las columnas
  orden = ["equipo_local", "goles_local", "goles_visitante", "equipo_visitante"]

  # Aplica el nuevo orden al DataFrame
  df = df[orden]

  return df
```

**FUNCIÓN** *calcular_puntos*:
1. Calcular: empates, ganados y perdidos con il/elif en numpy --> where del equipo local.
2. Hacer el mismo calcular del equipo visitante.
3. Creamos un dataframe para "locales" y otro para "visitantes" con los puntos.
4. Combinamos/juntamos las 2 tablas en una sola agrupando por equipo y sumando los puntos.
5. Devolvemos y guardamos el dataframe creado.

```python
def calcular_puntos(df):

  # Puntos local (3 si gana, 1 si empata, 0 si pierde)
  df["pts_local"] = np.where(df["goles_local"] > df["goles_visitante"], 3,
                    np.where(df["goles_local"] == df['goles_visitante'], 1, 0))

  # Puntos visitante (3 si gana, 1 si empata, 0 si pierde)
  df["pts_visitante"] = np.where(df["goles_visitante"] > df["goles_local"], 3,
                        np.where(df["goles_local"] == df["goles_visitante"], 1, 0))

  #creamos un dataframe con los puntos por equipo

  # primero creamos un dataframe por cada equipo: local y visitante
  locales = df[["equipo_local", "pts_local"]].rename(columns={"equipo_local": "equipo", "pts_local": "puntos"})
  visitantes = df[["equipo_visitante", "pts_visitante"]].rename(columns={"equipo_visitante": "equipo", "pts_visitante": "puntos"})

  #combinamos ambas tablas en una sola.
  df_puntos = (pd.concat([locales, visitantes]) #unimos las 2 tablas
            .groupby("equipo", as_index=False)["puntos"] # agrupamos por equipo
            .sum()  #sumamos los puntos
            .sort_values("puntos", ascending=False) #ordenamos de mayor a menos
            .reset_index(drop=True))

  return df_puntos
```

Sumamos los goles en contra y a favor de cada equipo.
1. Convertimos a "*int*" porque el valor de las columnas no es numérico.
2. Hacemos 2 dataframe: uno de locales y otro de visitantes. Renombramos las columnas segun los goles a favor o en contra.
3. Juntamos ambas tablas agrupando por equipo y sumando los goles.
4. Nos guarda y devuleve el dataframe.

```python
def calcular_goles(df):

  # Convertir a int, porque si no, acumula los resultados, no los suma.
  df["goles_local"]     = pd.to_numeric(df["goles_local"])
  df["goles_visitante"] = pd.to_numeric(df["goles_visitante"])


  # Goles como local
  locales = df[["equipo_local", "goles_local", "goles_visitante"]].rename(columns={
      "equipo_local":    "equipo",
      "goles_local":     "goles_favor",
      "goles_visitante": "goles_contra"
  })

  # Goles como visitante
  visitantes = df[["equipo_visitante", "goles_visitante", "goles_local"]].rename(columns={
      "equipo_visitante": "equipo",
      "goles_visitante":  "goles_favor",
      "goles_local":      "goles_contra"
  })

  # Concatenar y sumar
  df_goles = (pd.concat([locales, visitantes])
            .groupby("equipo", as_index=False)[["goles_favor", "goles_contra"]]
            .sum()
            .sort_values("goles_favor", ascending=False)
            .reset_index(drop=True))

  return df_goles
```

```python
#comprobamos
calcular_goles(df_historico)
```

**FUNCIÓN** *calcular_partidos*:

Calculamos los partidos ganados, empatados y perdidos.
1. Hacemos un dataframe solo con los equipos locales renombrando las columnas de los resultados.
2. Hacemos un dataframe solo con los equipos visitantes renombrando las columnas de los resultados.
3. Unimos las 2 tablas.
4. Establecemos las condiciones:
*   Si goles a favor es mayor que goles en contra --> suma 1 en la columna *ganados*.
*   Si goles a favor es igual que goles en contra --> suma 1 en la columna *empatados*.
*   Si goles a favor es menor que goles en contra --> suma 1 en la columna *perdidos*.

5. Agrupamos el dataframe por equipo, sumamos y ordenamos los resultados.
6. Devuelve el dataframe.

```python
def calcular_partidos(df):

  # hacemos un dataframe on los equipos locases
  locales = df[["equipo_local", "goles_local", "goles_visitante"]].copy()
  locales.columns = ["equipo", "favor", "contra"]

  # hacemos un df con los visitantes
  visitantes = df[["equipo_visitante", "goles_visitante", "goles_local"]].copy()
  visitantes.columns = ["equipo", "favor", "contra"]

  # Unimos los dos fd
  todos = pd.concat([locales, visitantes]).reset_index(drop=True)

  # utilizamos condicionales para calcular el resultado
  condiciones = [
      todos["favor"] > todos["contra"],
      todos["favor"] == todos["contra"],
      todos["favor"] < todos["contra"]
  ]

  todos["ganados"]   = np.select(condiciones, [1, 0, 0])
  todos["empatados"] = np.select(condiciones, [0, 1, 0])
  todos["perdidos"]  = np.select(condiciones, [0, 0, 1])

  # agrupamos el dataframe por equipo y summos los resutlados anteriores
  df_partidos = (todos.groupby("equipo", as_index=False)[["ganados", "empatados", "perdidos"]]
                .sum()
                .sort_values("ganados", ascending=False)
                .reset_index(drop=True))

  return df_partidos
```

**FUNCIÓN** *combinar_resultados*():
  
Juntamos todos los df en uno solo.
1. Recibe 3 dataframes.
2. Los combina en uno solo.
3. Ordenamos las columnas del dataframe par que siempre devuelva el mismo orden.
4. Devuelve el df final.

```python
def combinar_resultados(df1, df2, df3):
  df_clasificacion = df1 \
      .merge(df2, on="equipo", how="outer") \
      .merge(df3,    on="equipo", how="outer")
       #con outer, aparecen todos los equipos aunque no aparezcan en los tres dataframe.

  #ordenamos las columnas, porque según el orden de los df como variables, puede verse afectado.
  orden = ["equipo", "goles_favor", "goles_contra", "ganados", "perdidos", "empatados", "puntos"]
  df_clasificacion = df_clasificacion[orden]

  return df_clasificacion
```
