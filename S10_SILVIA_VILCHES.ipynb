**1. Consulta la API pública de JSONPlaceholder utilizando el método GET para obtener:**

* Lista de usuarios (/users)

```python
request_users = requests.get("https://jsonplaceholder.typicode.com/users")

print(request_users)
```

```python
users = request_users.json()
print(f"Lista de usuarios:{users}")
```

* Lista de tareas (/todos)

```python
request_todos = requests.get("https://jsonplaceholder.typicode.com/todos")

print(request_todos)
```

```python
todos = request_post.json()
print(f"La lista completa:{todos}")
```

**3. Realiza una solicitud a una publicación inexistente para obtener un error 404 y muestra el código de estado recibido.**


```python
cod_404 = requests.get(f"https://jsonplaceholder.typicode.com/posts/posts/567")

print(f"Código de estado recibido: {cod_404.status_code}")
```

💠 Como sabemos que hay 100 post, buscamos cualquier que sea superior a este número.

💠 El código 404 nos dice que no ha encontrado lo que buscamos ➡ La URL que queremos ver no existe.

**4. Realiza una solicitud POST para crear una nueva publicación ficticia. Incluye el título, el cuerpo del mensaje y un userId.**

* Mostrar la respuesta JSON.
* Mostrar el código de estado.




```python
new_post = {
    "title": "Hola, esta es mi publicación falsa.",
    "body": "El veloz murciélago hindú comía feliz cardillo y kiwi.",
    "userId": 26
}

resp_post = requests.post(f"https://jsonplaceholder.typicode.com/posts", json=new_post)

print("Respuesta JSON:", resp_post.json())
print("Código de estado:", resp_post.status_code)
```

💠Hemos creado una publicación donde seguimos el formato del diccionario.

💠 EL código 201 nos indica que el post se ha creado correctamente ➡ Creado con éxito.

**5. Realizar una solicitud PATCH para modificar parcialmente una entrada existente.**

* Mostrar la respuesta JSON.
* Mostrar el código de estado.

```python
update_post = {
    "title": "UPDATE PATCH"
}

resp_patch = requests.patch(f"https://jsonplaceholder.typicode.com/posts/4", json=update_post)

print("Respuesta JSON:", resp_patch.json())
print("Código de estado:", resp_patch.status_code)
```

💠 El código 200 nos dice que todo está “OK” ➡ La respuesta para una solicitud HTTP ha tenido éxito.

💠 Hemos modificado el titulo de la publicación con ID 4.

**6. Realizar una solicitud DELETE sobre una entrada.**

* Mostrar la respuesta JSON.
* Mostrar el código de estado.

```python
resp_delete = requests.delete(f"https://jsonplaceholder.typicode.com/posts/4")

print("Respuesta JSON:", resp_delete.json())
print("Código de estado:", resp_delete.status_code)
```

💠 El código 200 nos dice que todo está “OK” ➡ La respuesta para una solicitud HTTP ha tenido éxito.

💠 Eliminamos la entrada quehabiamos hecho.

```python
ajuntamentBCN = "https://opendata-ajuntament.barcelona.cat/data/api"
```

**1. Utiliza la API de datos abiertos de Barcelona para:**

* Buscar un conjunto de datos que te interese utilizando package_search y package_show.

💠 Buscamos datasets relacionados con el arbolado ("arbrat") de Barcelona.

```python
params = {"q": "arbrat", "rows": 10} # la letra q --> query, consulta.

resp_busqueda = requests.get(f"{ajuntamentBCN}/3/action/package_search", params=params)

print("Código de estado:", resp_busqueda.status_code)

resultado_busqueda = resp_busqueda.json()['result']
print(f"Número de datasets encontrados: {resultado_busqueda['count']}")
print()
for ds in resultado_busqueda['results']:
    print(f" - {ds['name']}  |  {ds['title']}")
```

**2. De entre los resultados obtenidos, selecciona uno que tenga recursos disponibles en formato CSV o JSON.**

💠 Elegimos el dataset *arbrat-zona* (árboles de zona de Barcelona) y consultamos su detalle para ver qué recursos tiene disponibles y cuáles permiten consulta vía *datastore_search*.

```python
resp_arboles = requests.get(f"{ajuntamentBCN}/3/action/package_show", params={"id": "arbrat-zona"})  #consultamos la info, toda la información del conjunto de datos llamado arbrat-zona

print("Código de estado:", resp_arboles.status_code)

data_arboles = resp_arboles.json()["result"]  # convertimos la respuesta en un diccionario.
print("---------------------------------------------------------")
print("Título del daraframe:", data_arboles["title"])
print("---------------------------------------------------------")
print("Recursos con datastore activo:")
for recurso in data_arboles["resources"]:  #buscamos, si el recurso nos permite hacer consultas con datastore_search, que nos muestre la info.
    if recurso.get("datastore_active"):
        print(f"- Nombre: {recurso['name']}")
        print(f"- Formato: {recurso['format']}")
        print(f"- Resource_id: {recurso['id']}")
```

```python
# Nos quedamos con el recurso CSV, que es el que tiene el datastore activo, que nos ha dado en la consulta anterior
resource_id = "29cd5c1f-11b1-404b-b3a5-ae29940b8c55" 
print("resource_id:", resource_id)
```

**3. Utilizando el resource_id del recurso seleccionado, realiza una consulta con datastore_search para:**

* Recuperar al menos 100 registros.

```python
params = {
    "resource_id": resource_id,
    "limit": 100
}

respuesta_ajt = requests.get(f"{ajuntamentBCN}/action/datastore_search", params=params)

print("Código de estado:", respuesta_ajt.status_code)

data = respuesta_ajt.json()['result']
print("---------------------------------------------------------")
print("Campos disponibles:", [f["id"] for f in data["fields"]])
print("---------------------------------------------------------")
print("Número de registros recuperados:", len(data["records"]))
```

**4. Convierte los resultados a un DataFrame.**


```python
df_arboles = pd.DataFrame(data['records'])

print("Forma del DataFrame:", df_arboles.shape)
df_arboles.head(5)
```

**5. Guarda el DataFrame en un archivo .csv.**

```python
df_arboles.to_csv("arbolest_zona_barcelona.csv", index=False, encoding="utf-8-sig")
print("Archivo guardado como - arbolest_zona_barcelona.csv - ")
```
